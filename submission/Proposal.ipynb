## Giới thiệu bài toán và dữ liệu

Website bán hàng có nhiều phiên truy cập kết thúc mà chưa mua. Đồ án khảo sát liệu hành vi đã tổng hợp theo phiên có đủ tín hiệu để xếp hạng các phiên theo khả năng mua, đồng thời đo sự đánh đổi giữa bỏ sót và báo nhầm.

**Câu hỏi nghiên cứu.** Trong năm mô hình phân loại cơ bản, mô hình nào cho Average Precision tốt nhất khi dự đoán một phiên UCI có dẫn đến mua hàng? Tại ngưỡng chọn trên train, mô hình phát hiện được bao nhiêu phiên mua và tạo bao nhiêu báo nhầm?

**Đơn vị quan sát** là một phiên truy cập website. **Đầu vào** gồm số trang xem, thời lượng, chỉ số thoát và ngữ cảnh phiên. **Đầu ra** là điểm mua và nhãn nhị phân. `Revenue=True` biểu thị phiên có mua hàng. Dữ liệu được tổng hợp theo phiên, nên phạm vi là đánh giá offline. [Nguồn UCI](https://doi.org/10.24432/C5F88Q)

## Số liệu và thiết kế

Dataset có **12.330 phiên**, 17 feature gốc và một nhãn. Tỷ lệ lớp mua là **15,47%**. Cấu hình chính dùng 16 feature và dành riêng `PageValues` cho phân tích độ nhạy vì thời điểm có sẵn chưa xác minh. [Mô tả biến](../data/README.md)

Chia train/test phân tầng 80/20 với seed 42. Khảo sát hành vi theo nhãn trên train. So sánh DummyClassifier, Logistic Regression, kNN, Decision Tree và Random Forest bằng cùng 5 fold phân tầng. Fit preprocessing trong mỗi fold. Chọn mô hình theo Average Precision và ngưỡng theo F1 của lớp mua trên dự đoán out-of-fold. Báo cáo Precision, Recall, F1, AP, ROC-AUC và confusion matrix. Dùng baseline luôn đoán không mua để giải thích hạn chế của Accuracy trên dữ liệu mất cân bằng.
